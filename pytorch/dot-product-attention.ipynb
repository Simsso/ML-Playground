```python
import torch
```

```python
class DotProductAttention(torch.nn.Module):
    def forward(self, k, q, v):
        """Computes dot product attention.
        
        Shapes (k, q): [batch_size, seq_len, kq_dim]
        Values: same but v_dim
        """

        attn_scores = torch.softmax(q @ k.transpose(-2, -1) / q.shape[-1] ** 0.5, dim=-1)
        # v: [b,l,v_dim]
        # scores: [b,l,l]
        weighted_vs = torch.matmul(attn_scores, v)  # [b,l,v_dim]
        return weighted_vs
```
